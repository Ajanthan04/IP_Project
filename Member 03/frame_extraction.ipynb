```python
import cv2
import os

# ─────────────────────────────────────────────
# PATHS
# ─────────────────────────────────────────────
VIDEO_PATH = "video/pothole1.mp4"
OUTPUT_FOLDER = "input_frames"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# ─────────────────────────────────────────────
# LOAD VIDEO
# ─────────────────────────────────────────────
cap = cv2.VideoCapture(VIDEO_PATH)

frame_count = 0
saved_count = 0

# Extract every N frames (adjust if needed)
FRAME_SKIP = 10   # change to 5 or 20 if needed

print("Extracting frames...")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    if frame_count % FRAME_SKIP == 0:
        filename = f"frame_{saved_count:04d}.jpg"
        path = os.path.join(OUTPUT_FOLDER, filename)

        cv2.imwrite(path, frame)
        saved_count += 1

    frame_count += 1

cap.release()

print(f"Done! Saved {saved_count} frames to '{OUTPUT_FOLDER}'")
```
