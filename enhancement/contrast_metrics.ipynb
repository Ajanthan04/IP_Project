```python
import cv2
import os
import numpy as np
import pandas as pd

original_folders = {
    "median": "denoised_median"
}
clahe_folders = {
    "median": "enhanced_median_clahe"
}

for method in original_folders.keys():
    orig_folder  = original_folders[method]
    clahe_folder = clahe_folders[method]
    contrast_data = []

    print(f"\nProcessing {method.upper()} frames...")

    for filename in sorted(os.listdir(orig_folder)):
        if not filename.endswith(".png"):
            continue

        orig_img  = cv2.imread(os.path.join(orig_folder,  filename), cv2.IMREAD_GRAYSCALE)
        clahe_img = cv2.imread(os.path.join(clahe_folder, filename), cv2.IMREAD_GRAYSCALE)

        if orig_img is None or clahe_img is None:
            continue

        orig_contrast  = np.std(orig_img)
        clahe_contrast = np.std(clahe_img)
        contrast_data.append([filename, orig_contrast, clahe_contrast])

    # ✅ Save metrics to CSV only — no comparison images needed
    csv_path    = f"contrast_values_{method}.csv"
    contrast_df = pd.DataFrame(contrast_data, columns=["Frame", "Original Contrast", "CLAHE Contrast"])
    contrast_df.to_csv(csv_path, index=False)  # ✅ fixed broken link
    print(f"{method.upper()} done. Contrast metrics saved to '{csv_path}'")

print("\nAll metrics saved successfully!")
```
