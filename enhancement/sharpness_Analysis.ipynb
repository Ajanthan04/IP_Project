```python
import cv2
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# INPUT FOLDERS (CLAHE outputs)
mc_input = 'enhanced_median_clahe'

# OUTPUT FOLDERS (Sharpened frames)
mc_output = 'SHARP_MC'

# Comparison image folder
comparison_folder = 'SHARP_COMPARISON'

# CSV output file
csv_output = 'sharpness_values.csv'

# Create folders if they don't exist
os.makedirs(mc_output, exist_ok=True)
os.makedirs(comparison_folder, exist_ok=True)

# Sharpening kernel
kernel = np.array([[0,-1,0],
                   [-1,5,-1],
                   [0,-1,0]])

results = []

# Function to measure sharpness
def sharpness_value(image):
    return cv2.Laplacian(image, cv2.CV_64F).var()

def process_images(input_folder, output_folder, label):

    for file in os.listdir(input_folder):

        if file.endswith(".jpg") or file.endswith(".png"):

            path = os.path.join(input_folder, file)

            # Read image directly as grayscale
            gray = cv2.imread(path, cv2.IMREAD_GRAYSCALE)

            # Slight smoothing before sharpening
            gaussian = cv2.GaussianBlur(gray, (5,5), 0)

            # Apply sharpening
            sharpened = cv2.filter2D(gaussian, -1, kernel)

            # Calculate sharpness values
            original_sharp = sharpness_value(gray)
            sharpen_sharp = sharpness_value(sharpened)

            # Save sharpened frame
            cv2.imwrite(os.path.join(output_folder, file), sharpened)

            # Create comparison figure
            plt.figure(figsize=(8,4))

            plt.subplot(1,2,1)
            plt.imshow(gray, cmap='gray')
            plt.title(f"Original\nSharpness={original_sharp:.2f}")
            plt.axis("off")

            plt.subplot(1,2,2)
            plt.imshow(sharpened, cmap='gray')
            plt.title(f"Sharpened\nSharpness={sharpen_sharp:.2f}")
            plt.axis("off")

            save_name = label + "_" + file
            plt.savefig(os.path.join(comparison_folder, save_name))
            plt.close()

            # Store results
            results.append([file, label, original_sharp, sharpen_sharp])

            print("Processed:", file)

# Run for both methods
process_images(mc_input, mc_output, "MC")


# Save sharpness values to CSV
df = pd.DataFrame(results, columns=["Frame","Method","Original Sharpness","Sharpened Sharpness"])
df.to_csv(csv_output, index=False)

print("All frames processed successfully!")
```
