```python
import cv2
import os
import numpy as np
import matplotlib.pyplot as plt

before_folder = "denoised_median"
after_folder = "enhanced_median_clahe"
output_folder = "histogram_comparisons"

os.makedirs(output_folder, exist_ok=True)

results = []

for filename in os.listdir(before_folder):
    if filename.endswith(".png"):

        before_path = os.path.join(before_folder, filename)
        after_path  = os.path.join(after_folder, filename)

        before = cv2.imread(before_path, cv2.IMREAD_GRAYSCALE)
        after  = cv2.imread(after_path,  cv2.IMREAD_GRAYSCALE)

        if before is None or after is None:
            continue

        # -------- Histogram --------
        hist_before = cv2.calcHist([before], [0], None, [256], [0,256])
        hist_after  = cv2.calcHist([after],  [0], None, [256], [0,256])

        # -------- Contrast --------
        contrast_before = np.std(before)
        contrast_after  = np.std(after)

        results.append((filename, contrast_before, contrast_after))

        # -------- Plot & Save --------
        plt.figure(figsize=(8,4))

        plt.subplot(1,2,1)
        plt.title("Before")
        plt.plot(hist_before)

        plt.subplot(1,2,2)
        plt.title("After CLAHE")
        plt.plot(hist_after)

        plt.tight_layout()

        save_path = os.path.join(output_folder, filename)
        plt.savefig(save_path)
        plt.close()

# -------- Summary --------
print("\nFrame-wise contrast comparison:\n")
for r in results[:10]:   # print first 10 only
    print(f"{r[0]} | Before: {r[1]:.2f} | After: {r[2]:.2f}")

print("\nAll histogram comparison images saved!")
```
