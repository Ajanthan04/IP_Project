```python
import cv2
import matplotlib.pyplot as plt
import os

frames_folder     = "frames"
gaussian_folder   = "denoised_gaussian"
median_folder     = "denoised_median"
comparison_folder = "comparison_results"
hist_folder       = "histogram_results"

os.makedirs(comparison_folder, exist_ok=True)
os.makedirs(hist_folder, exist_ok=True)

frame_files = sorted([f for f in os.listdir(frames_folder) if f.endswith(".png")])

for idx, file in enumerate(frame_files[:30]):
    frame_number = file.split("_")[1].split(".")[0]

    original = cv2.imread(os.path.join(frames_folder, file), cv2.IMREAD_GRAYSCALE)
    gaussian = cv2.imread(os.path.join(gaussian_folder, file), cv2.IMREAD_GRAYSCALE)
    median   = cv2.imread(os.path.join(median_folder, file), cv2.IMREAD_GRAYSCALE)

    # ✅ resize fix — makes all images same size
    original = cv2.resize(original, (640, 480))
    gaussian = cv2.resize(gaussian, (640, 480))
    median   = cv2.resize(median,   (640, 480))

    plt.figure(figsize=(12, 4))
    plt.subplot(1, 3, 1)
    plt.imshow(original, cmap='gray')
    plt.title(f"Original - Frame {frame_number}")
    plt.axis('off')
    plt.subplot(1, 3, 2)
    plt.imshow(gaussian, cmap='gray')
    plt.title(f"Gaussian - Frame {frame_number}")
    plt.axis('off')
    plt.subplot(1, 3, 3)
    plt.imshow(median, cmap='gray')
    plt.title(f"Median - Frame {frame_number}")
    plt.axis('off')
    plt.tight_layout()
    plt.savefig(os.path.join(comparison_folder, f"comparison_frame_{frame_number}.png"))
    plt.close()  # ✅ fixed broken link

    plt.figure(figsize=(12, 4))
    plt.subplot(1, 3, 1)
    plt.hist(original.ravel(), bins=256, range=[0, 256], color='black')
    plt.title(f"Original Histogram - Frame {frame_number}")
    plt.subplot(1, 3, 2)
    plt.hist(gaussian.ravel(), bins=256, range=[0, 256], color='blue')
    plt.title(f"Gaussian Histogram - Frame {frame_number}")
    plt.subplot(1, 3, 3)
    plt.hist(median.ravel(), bins=256, range=[0, 256], color='green')
    plt.title(f"Median Histogram - Frame {frame_number}")
    plt.tight_layout()
    plt.savefig(os.path.join(hist_folder, f"histogram_frame_{frame_number}.png"))
    plt.close()  # ✅ fixed broken link

print("✅ Comparison images and histograms for 30 frames are ready!")
print(f"Check '{comparison_folder}' and '{hist_folder}' folders for results.")
```
